# Técnica Smote para balancear os dados
Dados disponiveis em: [credit_simple.csv](https://github.com/TakashiNishii/CursoIA/blob/main/credit_simple.csv)

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder
from imblearn.over_sampling import SMOTENC
```

```python
credito = pd.read_csv("credit_simple.csv", sep=";")
credito.head()
```

```python
count = credito.groupby(['CLASSE']).size()
count
```

```python
# Iremos criar 400 dados sinteticos de classe ruim para balancear
y = credito['CLASSE'].values
X = credito.iloc[:,:-1].values
```

```python
labelencoder = LabelEncoder()
for i in range(X.shape[1]):
  if X[:,i].dtype == 'object':
    X[:,i] = labelencoder.fit_transform(X[:,i])
```

```python
sm = SMOTENC(random_state=0, categorical_features=[3,5,6])
X_res, y_res = sm.fit_resample(X, y)
```

```python
X_res
```

```python
df = pd.DataFrame({'CLASSE': y_res})
df.value_counts()
```
